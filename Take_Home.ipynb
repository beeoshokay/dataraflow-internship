```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")


class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, interest_rate=1.71):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = self.balance * self.interest_rate
        self.balance += interest
        print(f"Interest added: {interest}. New balance: {self.balance}")


acc = SavingsAccount("Blessing", 1000, 0.08)
acc.deposit(5830)
acc.withdraw(240)
acc.add_interest()
```

```python
# Q1: The two methods needed to implement an iterator are: __iter__() and __next__()
# Q2: The statement used to stop iteration is: raise StopIteration
# Q3: True. lists, tuples, dictionaries, and sets are all iterable objects because each one implements the iterator protocol internally, which lets iter() pull an iterator out of them.

# Manual iteration over a list
numbers = [10, 20, 30, 40, 50]
my_iter = iter(numbers)

while True:
    try:
        value = next(my_iter)
        print(value)
    except StopIteration:
        print("No more items in the list.")
        break


# CountDown class
class CountDown:
    def __init__(self, n):
        self.n = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.n >= 1:
            current = self.n
            self.n -= 1
            return current
        else:
            raise StopIteration


for num in CountDown(5):
    print(num)
```

```python
class FileLineReader:
    def __init__(self, filename):
        self.filename = filename

    def __iter__(self):
        self.file = open(self.filename, "r")
        return self

    def __next__(self):
        line = self.file.readline()
        if line:
            return line.strip()
        else:
            self.file.close()
            raise StopIteration


with open("sample.txt", "w") as f:
    f.write("Break\nYour\nElastic Limit\n")

for line in FileLineReader("sample.txt"):
    print(line)
```

```python
def show_len_behavior():
    my_list = [2, 6, 9, 12, 16]

    try:
        print("Length of list using built-in len():", len(my_list))
    except UnboundLocalError as e:
        print("Error:", e)
        print("Explanation: Python saw that 'len' gets assigned later in")
        print("this function, so it treats 'len' as a LOCAL variable for")
        print("the whole function — even on this line, before the")
        print("assignment happens. This is why it fails immediately,")
        print("instead of only failing after 'len' is reassigned.")

    len = 100
    print("len is now:", len)

show_len_behavior()
```

Create an iterator class `EvenNumbers` that takes a limit and iterates over even numbers only. Use it in a `for` loop.

```python
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit

    def __iter__(self):
        self.n = 0
        return self

    def __next__(self):
        if self.n <= self.limit:
            result = self.n
            self.n += 4
            return result
        else:
            raise StopIteration

for num in EvenNumbers(36):
    print(num)
```

Create two classes: `Flyer` (method `move()` returns `"Flying"`) and `Swimmer` (method `move()` returns `"Swimming"`).  
Create a class `Duck` inheriting from both.  
- Use `super()` to decide which `move()` executes.  
- Explain the **Method Resolution Order (MRO)**.  

```python
class Flyer:
    def move(self):
        return "Flying"

class Swimmer:
    def move(self):
        return "Swimming"

class Duck(Flyer, Swimmer):
    def move(self):
        # super() follows the Method Resolution Order (MRO):
        # Duck -> Flyer -> Swimmer -> object
        return super().move()

d = Duck()
print(d.move())
print(Duck.__mro__)
```

Method Resolution Order is the order that Python uses to resolve a method name lookup in the case where a class inherits from multiple parent classes. It goes left to right through the classes listed in the class Duck(Flyer, Swimmer): line so super().move() here calls Flyer's version because Flier is first.

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

```python
class Fibonacci:
    def __init__(self, n):
        self.n = n

    def __iter__(self):
        self.a, self.b = 0, 1
        self.count = 0
        return self

    def __next__(self):
        if self.count < self.n:
            result = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return result
        else:
            raise StopIteration

for num in Fibonacci(11):
    print(num)
```

## Scenario

### 5. Extra Features

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

    def get_description(self):
        return f"{self.title} by {self.author} ({self.year})"


class EBook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        return f"{super().get_description()} - EBook, {self.format}, {self.file_size}MB"


class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return f"{super().get_description()} - Printed, {self.weight}g, {self.dimensions}"


class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = []  # protected

    def borrow_book(self, book):
        raise NotImplementedError

    def get_borrowed_books(self):
        return self._borrowed_books

    def __iter__(self):
        return iter(self._borrowed_books)


class StudentMember(Member):
    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level
        self.borrow_limit = 5

    def borrow_book(self, book):
        if len(self._borrowed_books) < self.borrow_limit:
            self._borrowed_books.append(book)
            return True
        return False


class TeacherMember(Member):
    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department
        self.borrow_limit = 10

    def borrow_book(self, book):
        if len(self._borrowed_books) < self.borrow_limit:
            self._borrowed_books.append(book)
            return True
        return False


class Library:
    total_borrowed = 0  # class-level variable

    def __init__(self, name):
        self.__library_name = name  # private
        self._books = []
        self._members = []

    def get_library_name(self):
        return self.__library_name

    def set_library_name(self, name):
        self.__library_name = name

    def add_book(self, book):
        self._books.append(book)

    def add_member(self, member):
        self._members.append(member)

    def search_by_title(self, title):
        return [b for b in self._books if title.lower() in b.title.lower()]

    def search_by_author(self, author):
        return [b for b in self._books if author.lower() in b.author.lower()]

    def lend_book(self, member, book):
        if member.borrow_book(book):
            Library.total_borrowed += 1
            print(f"{member.name} borrowed '{book.title}'")
        else:
            print(f"{member.name} has reached their borrowing limit")

    def __iter__(self):
        return iter(self._books)


# --- Demonstration ---
library = Library("National Library Of Nigeria")

book1 = EBook(
    "Textbook of Medical Physiology",
    "John E. Hall, Michael E. Hall",
    2020,
    45,
    "PDF"
)

book2 = PrintedBook(
    "Robbins and Cotran Pathologic Basis of Disease",
    "Vinay Kumar, Abul K. Abbas, Jon C. Aster",
    2020,
    3000,
    "10th Ed."
)

book3 = PrintedBook(
    "Tell Me Your Dreams",
    "Sidney Sheldon",
    1998,
    350,
    "Paperback"
)

book4 = PrintedBook(
    "Criminal Law",
    "Jonathan Herring",
    2020,
    900,
    "11th Ed."
)

book5 = PrintedBook(
    "Sophie's World",
    "Jostein Gaarder",
    1991,
    500,
    "Paperback"
)

library.add_book(book1)
library.add_book(book2)
library.add_book(book3)
library.add_book(book4)
library.add_book(book5)

student = StudentMember("Blessing", "U2016/470054", 300)
teacher = TeacherMember("Dr Lekpa", "T011", "Human Anatomy")
student2 = StudentMember("Olivia", "U2018/470099", 300)

library.add_member(student)
library.add_member(teacher)
library.add_member(student2)

library.lend_book(student, book1)
library.lend_book(teacher, book2)
library.lend_book(student, book3)

print("\nAll books in library:")
for book in library:
    print(book.get_description())

print("\nBooks borrowed by student:")
for book in student:
    print(book.get_description())

print("\nBooks borrowed by Olivia:")
for book in student2:
    print(book.get_description())

print("\nTotal books borrowed across library:", Library.total_borrowed)
print("Library name:", library.get_library_name())

print("\nSearch by title 'Robbins':", [b.title for b in library.search_by_title("Robbins")])
```

The general idea of this video is just different ways the same data can be arranged depending on what it's needed for.

Certain types like Tuples become immutable once created, meaning they can't be changed, and others like lists are changeable after being typed in. Small builtins like sorted(), min(), max(), count() do more work than the few characters they take to type.

Sets do not allow for duplicates. Dictionaries map one thing to another; the same energy as the shopping cart exercise (item → price).

A stack follows LIFO (Last In, First Out)i.e. the most recently added item is the first one removed, using append() to push and pop() to remove. A queue follows FIFO (First In, First Out) where items are removed in the order they were added. A heap orders elements by priority rather than insertion order, so the highest (or lowest) priority element is always accessed first, regardless of when it was added.

A linked list stores elements as nodes, each holding a value and a pointer to the next node's location. A doubly linked list adds a pointer back to the previous node for twoway visit order. A circular linked list connects the last node back to the first, removing the fixed endpoint.

A binary search tree adds an ordering rule to this structure: smaller values go left, larger values go right. The visit order method decides the order.

A graph drops the hierarchy order, letting any node connect to any other through an edge. It's represented either as an adjacency list (each node mapped to its connections) or an adjacency matrix (a grid marking which nodes connect).

Overall, this video reinforced that data structures aren't interchangeable. They each impose a specific rule on how data is accessed, ordered, or connected, and the right choice depends on the shape of the problem rather than personal familiarity with any single structure.
